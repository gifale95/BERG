## Print out folders

```python
import os

def print_tree(root_path, ignore=None, prefix=""):
    if ignore is None:
        ignore = {
            ".DS_Store",
            "__pycache__",
            ".git",
            ".gitignore",
            ".Trash",
            ".idea",
            ".vscode"
        }

    try:
        entries = sorted(os.listdir(root_path))
    except PermissionError:
        print(prefix + "⚠️ [Permission Denied]")
        return

    entries = [e for e in entries if e not in ignore and not e.startswith("._")]

    for i, entry in enumerate(entries):
        path = os.path.join(root_path, entry)
        is_last = i == len(entries) - 1

        connector = "└── " if is_last else "├── "
        print(prefix + connector + entry)

        if os.path.isdir(path):
            extension = "    " if is_last else "│   "
            print_tree(path, ignore, prefix + extension)


# Example usage:
print_tree("/Volumes/ExtremeSSD/Datasets/Wingert_ferret")
```

```python
import os
import numpy as np

def list_rois(metadata_dir, subjects=['UTS01', 'UTS02', 'UTS03']):
    all_rois = {}
    for sub in subjects:
        meta = np.load(os.path.join(metadata_dir, f'metadata_{sub}.npy'), allow_pickle=True).item()
        rois = sorted(meta['roi'].keys())
        all_rois[sub] = set(rois)
        print(f'{sub}: {len(rois)} ROIs — {rois}')
    
    common = set.intersection(*all_rois.values())
    print(f'\nCommon ROIs ({len(common)}): {sorted(common)}')
    
    for sub in subjects:
        unique = all_rois[sub] - common
        if unique:
            print(f'  {sub} only: {sorted(unique)}')

list_rois('/Volumes/ExtremeSSD/brain-encoding-response-generator/encoding_models/modality-fmri/train_dataset-lebel2023/model-opt_1_3b_ridge/metadata')
```
